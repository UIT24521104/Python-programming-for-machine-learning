```python
import pandas as pd
import numpy as np

np.random.seed(42)
n = 1000

data = pd.DataFrame({
    "Age": np.random.randint(20, 70, n),
    "Education": np.random.randint(10, 20, n),         # years of education
    "Income": np.random.normal(5000, 1500, n)          # monthly income
})

# =========================
# Tạo MAR missing cho `Income`
# Quy tắc MAR (ví dụ):
# - Người có Education thấp (<=13) có khả năng bỏ khai Income cao hơn
# - Người lớn tuổi (>50) cũng có khả năng bỏ khai Income cao hơn
# => missing xác suất phụ thuộc vào các biến quan sát được (Age, Education) => MAR
# =========================
rng = np.random.RandomState(42)

base_prob = 0.05
edu_effect = 0.25   # tăng xác suất missing nếu Education <= 13
age_effect = 0.10   # tăng xác suất missing nếu Age > 50

p_missing = (
    base_prob
    + edu_effect * (data["Education"] <= 13).astype(float)
    + age_effect * (data["Age"] > 50).astype(float)
)

# đảm bảo xác suất không vượt 1
p_missing = np.clip(p_missing, 0, 1)

mask = rng.rand(n) < p_missing

data_missing = data.copy()
data_missing.loc[mask, "Income"] = np.nan

# =========================
# Lưu dữ liệu
# =========================
data.to_csv("dataset_goc.csv", index=False, encoding="utf-8")
data_missing.to_csv("dataset_missing.csv", index=False, encoding="utf-8")

# Thông tin nhanh
print("Saved dataset_goc.csv and dataset_missing.csv")
print("\nMissing counts (dataset_missing):")
print(data_missing.isnull().sum())
print("\nMissing fraction (dataset_missing):")
print(data_missing.isnull().mean())
```
