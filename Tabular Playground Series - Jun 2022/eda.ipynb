```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv(r'.\data.csv')
```

```python

f4_cols = [col for col in data.columns if col.startswith('F_4')]

plt.figure(figsize=(10, 8))
sns.heatmap(data[f4_cols].corr(method='spearman'), cmap='coolwarm', center=0)
plt.title("Spearman Correlation - Nhóm F_4")
plt.show()

missing_f4_per_row = data[f4_cols].isnull().sum(axis=1)
print("--- Phân bố số lượng ô khuyết F_4 trên mỗi hàng ---")
print(missing_f4_per_row.value_counts().sort_index())

print("\n--- Phương sai trung bình của từng nhóm ---")
for i in range(1, 5):
    cols = [col for col in data.columns if col.startswith(f'F_{i}')]
    if cols:
        print(f"Nhóm F_{i}: {data[cols].var().mean():.4f}")
```
